```python
import pandas as pd 
Stud_detail = {
     "RollNo": [100,101,102,103,104,105,106,107],
     "Name": ["Deepak","Divyansh","Ashish","Arman","Jabir","Amit","Deepanshi","Sanjana"],
     "Course": ["Python","Java","Fullstuck","Python","Java","Fullstuck","Python","Java"],
     "City": ["Delhi","Mumbai","Pune","Delhi","Lucknow","Mumbai","Gujrat","Delhi"],
     "Age": [21,22,23,20,19,21,24,20],
     "Marks": [70,78,87,85,98,90,93,75]
}
 
df = pd.DataFrame(Stud_detail)

print(df) 
```

```python
# Display only name column.
# Display Name and Marks.

df = pd.DataFrame(Stud_detail)
print(df["Name"])
print(df[["Name","Marks"]])    
```

```python
# Display the first five row.

df = pd.DataFrame(Stud_detail)
print(df.head())
```

```python
# Display the first row.

df = pd.DataFrame(Stud_detail)
print(df.head(1))
```

```python
# Display Students with marks greater than 80.

df = pd.DataFrame(Stud_detail)
result = df[df["Marks"]>80]
print(result)
```

```python
# Display students whose age is greater than 20.

df = pd.DataFrame(Stud_detail)
result = df[df["Age"]>21]
print(result)
```

```python
# Display students from the python course.

df = pd.DataFrame(Stud_detail)
# print(df["Course"])
print(df[df["Course"] == "Python"])
```

```python
# Display students from delhi.

df = pd.DataFrame(Stud_detail)
print(df[df["City"] == "Delhi"])
```

```python
# Display students with marks greater than 80 and age greater than 20.

df = pd.DataFrame(Stud_detail)
result = df[(df["Marks"]>80) & (df["Age"]>20)]
print(result)
```

```python
# Display students from either python or java.

df = pd.DataFrame(Stud_detail)
print(df[df["Course"].isin(["Python","Java"])])
```
